# Food delivery operations and customer analytics — EDA

Dataset: [deepeshkansotia/food-delivery-operations-and-customer-analytics](https://www.kaggle.com/datasets/deepeshkansotia/food-delivery-operations-and-customer-analytics).

**Goal:** Characterize orders (timing, geography, customer traits), inspect operational signals (traffic, weather, ratings, fees), and relate them to **actual delivery duration**.

**CSV resolution:** Under `/kaggle/input`, local `data/`, or a `kagglehub` download directory, pick the `.csv` whose header contains **all** `EXPECTED_COLS`; if several match, pick the **largest file by byte size**.


```python
from __future__ import annotations

from IPython import get_ipython

_ip = get_ipython()
if _ip is not None:
    _ip.run_line_magic("matplotlib", "inline")

import kagglehub
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from pathlib import Path

from IPython.display import display

sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (11, 5)

DATASET_SLUG = "deepeshkansotia/food-delivery-operations-and-customer-analytics"
EXPECTED_COLS = [
    "order_id",
    "city_tier",
    "customer_age",
    "customer_loyalty_score",
    "order_hour",
    "order_day_of_week",
    "order_month",
    "delivery_distance_km",
    "preparation_time_minutes",
    "delivery_time_minutes",
    "estimated_delivery_time",
    "traffic_level_score",
    "weather_severity_score",
    "restaurant_rating",
    "delivery_partner_rating",
    "customer_rating",
    "order_value",
    "delivery_fee",
    "discount_amount",
    "tip_amount",
    "final_amount_paid",
    "number_of_items",
    "cancellation_flag",
    "delayed_delivery_flag",
    "refund_flag",
    "promo_code_used",
    "premium_customer_flag",
    "festival_or_weekend_flag",
    "delivery_partner_experience_years",
    "delivery_efficiency_score",
]


def _headers_match(path: Path) -> bool:
    peek = pd.read_csv(path, nrows=0)
    return set(EXPECTED_COLS) <= set(peek.columns)


def resolve_csv_path() -> Path:
    roots: list[Path] = []
    kaggle_root = Path("/kaggle/input")
    if kaggle_root.exists():
        roots.extend(sorted(kaggle_root.iterdir()))
    local_data = Path("data")
    if local_data.exists():
        roots.append(local_data)
    for root in roots:
        candidates = [p for p in root.rglob("*.csv") if p.is_file() and _headers_match(p)]
        if candidates:
            return max(candidates, key=lambda p: p.stat().st_size)
    hub = Path(kagglehub.dataset_download(DATASET_SLUG))
    candidates = [p for p in hub.rglob("*.csv") if p.is_file() and _headers_match(p)]
    assert candidates, f"no CSV under {hub} with columns {EXPECTED_COLS}"
    return max(candidates, key=lambda p: p.stat().st_size)


csv_path = resolve_csv_path()
print("CSV:", csv_path)

df = pd.read_csv(csv_path, low_memory=False)
missing_cols = set(EXPECTED_COLS) - set(df.columns)
assert not missing_cols, f"missing columns: {sorted(missing_cols)}"
extra = [c for c in df.columns if c not in EXPECTED_COLS]
assert not extra, f"unexpected columns: {extra}"
df = df[EXPECTED_COLS]

assert len(df) == 15_000
assert df["order_id"].nunique() == 15_000

print(len(df), "rows ×", df.shape[1], "cols")
miss = df.isna().sum()
if miss.any():
    print("Missing values (columns > 0):")
    print(miss[miss > 0])

display(df.sample(5, random_state=7))
display(df.dtypes)
display(df.describe(include=[np.number]).T)
```

## Univariate analysis

**Discrete timing / geography:** `city_tier`, `order_day_of_week`, `order_month` — counts and bar charts.

**Continuous:** core timing and logistics (`customer_age`, loyalty, hour, distance, prep, realized delivery) plus operational / commercial numerics when informative.

**Boolean flags:** quick prevalence bars.


```python
DISCRETE = ["city_tier", "order_day_of_week", "order_month"]

for col in DISCRETE:
    vc = df[col].value_counts().sort_index()
    plt.figure(figsize=(10, 4))
    sns.barplot(x=vc.index.astype(str), y=vc.values, hue=vc.index.astype(str), palette="muted", legend=False)
    plt.xlabel(col)
    plt.ylabel("count")
    plt.title(f"{col} — value counts")
    plt.tight_layout()
    plt.show()

CONTINUOUS_MAIN = [
    "customer_age",
    "customer_loyalty_score",
    "order_hour",
    "delivery_distance_km",
    "preparation_time_minutes",
    "delivery_time_minutes",
]
CONTINUOUS_EXTRA = [
    "estimated_delivery_time",
    "traffic_level_score",
    "weather_severity_score",
    "restaurant_rating",
    "delivery_partner_rating",
    "customer_rating",
    "order_value",
    "delivery_fee",
    "discount_amount",
    "tip_amount",
    "final_amount_paid",
    "delivery_partner_experience_years",
    "delivery_efficiency_score",
]

fig, axes = plt.subplots(2, 3, figsize=(14, 8))
axes = axes.flatten()
for ax, col in zip(axes, CONTINUOUS_MAIN):
    sns.histplot(df[col].dropna(), kde=True, ax=ax)
    ax.set_title(col)
plt.suptitle("Core continuous distributions", y=1.02)
plt.tight_layout()
plt.show()

fig, axes = plt.subplots(4, 4, figsize=(16, 14))
axes = axes.flatten()
for ax, col in zip(axes, CONTINUOUS_EXTRA):
    sns.histplot(df[col].dropna(), kde=True, ax=ax)
    ax.set_title(col, fontsize=9)
for ax in axes[len(CONTINUOUS_EXTRA) :]:
    ax.axis("off")
plt.suptitle("Additional numeric fields", y=1.01)
plt.tight_layout()
plt.show()

bool_cols = list(df.select_dtypes(include=bool).columns)
ncols = 3
nrows = int(np.ceil(len(bool_cols) / ncols))
fig, axes = plt.subplots(nrows, ncols, figsize=(12, 3 * nrows))
axes = np.atleast_2d(axes)
for idx, col in enumerate(bool_cols):
    r, c = divmod(idx, ncols)
    ax = axes[r, c]
    vc = df[col].value_counts().sort_index()
    sns.barplot(x=vc.index.astype(str), y=vc.values, ax=ax, hue=vc.index.astype(str), palette="pastel", legend=False)
    ax.set_title(col)
    ax.set_xlabel("")
for idx in range(len(bool_cols), nrows * ncols):
    r, c = divmod(idx, ncols)
    axes[r, c].axis("off")
plt.suptitle("Boolean flags — prevalence", y=1.02)
plt.tight_layout()
plt.show()
```

### Interpretation (univariate)

- **Orders are evenly spread across tiers, weekdays, and months** in this synthetic-style panel (use plots above to confirm).
- **Delivery times are right-skewed** toward longer durations while **distance** spans the operational range uniformly — tail behavior matters for SLA dashboards.
- **Missing ratings / tips** (if any) concentrate in a small subset of rows — multivariate plots treat NaNs via `.dropna()`.


## Multivariate analysis

- Pearson correlations among numeric fields (booleans encoded as 0/1).
- **Delivery duration vs distance** (density).
- **Delivery duration by city tier**.
- **Mean delivery time** by **order hour × city tier**.
- **Delayed delivery** vs **city tier** (two categorical).


```python
num_df = df.drop(columns=["order_id"]).copy()
corr = num_df.corr(numeric_only=True)

plt.figure(figsize=(14, 12))
sns.heatmap(corr, cmap="vlag", center=0, square=False)
plt.title("Pearson correlation — numeric features")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 6))
hb = plt.hexbin(
    df["delivery_distance_km"],
    df["delivery_time_minutes"],
    gridsize=35,
    cmap="viridis",
    mincnt=1,
)
plt.colorbar(hb, label="count")
plt.xlabel("delivery_distance_km")
plt.ylabel("delivery_time_minutes")
plt.title("Delivery duration vs distance (hexbin)")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="city_tier", y="delivery_time_minutes", hue="city_tier", palette="muted", legend=False)
plt.title("delivery_time_minutes by city_tier")
plt.tight_layout()
plt.show()

pivot = df.pivot_table(
    values="delivery_time_minutes",
    index="order_hour",
    columns="city_tier",
    aggfunc="mean",
)
plt.figure(figsize=(10, 8))
sns.heatmap(pivot, cmap="rocket_r", linewidths=0.5)
plt.title("Mean delivery_time_minutes: order_hour × city_tier")
plt.tight_layout()
plt.show()

ct = pd.crosstab(df["city_tier"], df["delayed_delivery_flag"], normalize="index") * 100
plt.figure(figsize=(7, 4))
sns.heatmap(ct, annot=True, fmt=".1f", cmap="Blues")
plt.title("delayed_delivery_flag (% within city_tier)")
plt.ylabel("city_tier")
plt.xlabel("delayed_delivery_flag")
plt.tight_layout()
plt.show()

target = "delivery_time_minutes"
assoc = corr[target].drop(labels=[target]).abs().sort_values(ascending=False)
print("Top |corr| with delivery_time_minutes:")
print(assoc.head(12).to_string())
```

## Takeaways (agent-readable)

- **Schema:** 15k orders, unique `order_id`; **30** columns spanning logistics, ratings, pricing, flags, and efficiency scores — several numerics drive **delivery_time_minutes** (see correlation heatmap and ranked list).
- **Distance vs duration:** Hexbin shows whether longer routes systematically inflate realized delivery times (nonlinear saturation would imply congestion or routing effects).
- **City tier:** Boxplots and the hour×tier heatmap reveal whether certain tiers or peak hours inflate average durations.
- **Delayed flag:** Crosstab-normalized heatmap compares delay prevalence across tiers.
- **Data quality:** Up to three columns carry sparse NaNs (partner/customer ratings, tips); analyses use `.dropna()` where needed.
